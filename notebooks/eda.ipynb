# 10Pearls AQI Predictor EDA

This notebook investigates AQI trends, distributions, missingness, and relationships with weather variables.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

DATA_PATH = Path('../data/processed/training_data.csv')
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
df.shape
```

```python
df[['date', 'aqi', 'pm25', 'pm10', 'temperature', 'humidity', 'pressure', 'wind_speed']].describe().T
```

```python
missing = df.isna().mean().sort_values(ascending=False)
missing[missing > 0]
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df['date'], df['aqi'], color='#d1495b')
plt.title('Daily AQI Trend')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.grid(alpha=0.25)
plt.show()
```

```python
df[['aqi', 'pm25', 'pm10', 'temperature', 'humidity']].hist(figsize=(12, 8), bins=15)
plt.suptitle('AQI and Weather Distributions')
plt.tight_layout()
plt.show()
```

```python
correlations = df.select_dtypes('number').corr()['aqi'].sort_values(ascending=False)
correlations
```
